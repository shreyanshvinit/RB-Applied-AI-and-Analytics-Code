```python
password = input("Enter a password: ")
pass_len = len(password) >= 8
pass_char = '@' in password
strong = pass_len and pass_char
print(f"Password: {password}")
print(f"Length OK (>= 8): {pass_len}")
print(f"Contains '@': {pass_char}")
print(f"Strong: {strong}")
```
