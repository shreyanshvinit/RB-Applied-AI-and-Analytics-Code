## Track B - Text Article Analyzer
 A python program designed to analyze any article or text file and provide a short analytical summary. The program output is a clean and formatted summary containing:

* Word count
* Sentence count
* Average word length
* Estimated reading time
* Most common used words

```python
#Function1: for importing and reading the article

def import_article(filename):
  try:
    with open(filename, "r") as file:
      text = file.read()
      return text

  except FileNotFoundError:
    print("File not found. Please check the file name and try again.")
    return None
```

```python
#Function2: for counting total words

def word_count(text):
  return len(text.split())
```

```python
#Function4: for average word length

def avg_word_len(text):
  words = text.split()

  #removing punctuations from words
  punctuation = r'!"#$%&\'()*+,-./:;<=>?@[]^_`{|}\~'

  cleaned_words = []
  for word in words:
    cleaned_word = word.strip(punctuation)
    if cleaned_word:
      cleaned_words.append(cleaned_word)

  if not cleaned_words:
    return 0

  total_length = sum(len(word) for word in cleaned_words)
  return total_length / len(cleaned_words)
```

```python
#Function5: Estimating reading time

def reading_time(text):
  word = word_count(text)
  words_read_per_minute = 200
  return (word / words_read_per_minute)
```

```python
#Function6: finding most commonly used words and formatting output
def common_word(text,top_n=10):
    words = text.lower().split()
    punctuation = r'!"#$%&\'()*+,-./:;<=>?@[\]^_{|}~'
    filler_words = {'the', 'and', 'is', 'a', 'of', 'in', 'to', 'for', 'etc.', 'that', 'them', 'by', 'their'}
    word_count = {}

    for word in words:
        word = word.strip(punctuation)

        #counting words frequency, ignoring filler words
        if word and word not in filler_words:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1

    #sorting words in decreasing order of frequency
    word_sort = sorted(word_count.items(), key=lambda x: x[1], reverse=True)
    top_words = word_sort[:top_n]

    return top_words
```

```python
#Function7: the analyzer output
def text_analysis(text):
    print("\n=== ARTICLE ANALYSIS ===")

    print(f"{"Word count":<25}: {word_count(text)}")

    print(f"{"Sentence count":<25}: {sentence_count(text)}")

    print(f"{"Average word length":<25}: {round(avg_word_len(text), 2)} characters")


    #Reading time
    time = reading_time(text)

    if time < 1:
        print(f"{"Estimated reading time":<25}: Less than 1 minute")
    else:
        print(f"{"Estimated reading time":<25}: ~{round(time, 1)} minutes")

    #Most common words
    print(f"Top 10 most common words:")

    for i, (word, count) in enumerate(common_word(text), start=1):
        print(f"  {i:>2}. {word:<12} ({count})")
    print("   ...")
```

```python
#Main function
def analyse(filename):

    #import article
    article = import_article(filename)

    # Check if article was loaded
    if article is None:
        return

    # Check if article is empty
    if article.strip() == "":
        print("Error: .txt file is empty.")
        return

    # Start analysis
    text_analysis(article)
```
